```python
# 19. Find the second-largest number.
numbers=[67, 5, 34, 12, 89, 45, 23, 7, 45, 12]
largest_num=numbers[0]
second_largest_num=numbers[0]
for i in numbers:
    if largest_num<i:
        largest_num=i
    elif i>second_largest_num and i!=largest_num:
        second_largest_num=i
print(second_largest_num)
```

```python
# 20. Find the second-smallest number.
numbers=[67, 5, 34, 12, 89, 45, 23, 7, 45, 12]
smallest=numbers[0]
Second_smallest=numbers[0]
for i in numbers:
    if i < smallest:
        smallest=i
    if i < Second_smallest and i != smallest:
        Second_smallest=i
print(Second_smallest)
```

### Challenging Tasks

```python
# 36. Merge two lists without using the + operator.
list1 = [1, 2, 3]
list2 = [4, 5, 6]
merged = list1.copy()
merged.extend(list2)
print(merged)
```

```python
# 37. Find common elements between two lists.
list1 = [10, 20, 30, 40]
list2 = [30, 40, 50, 60]
common = []
for item in list1:
    if item in list2 and item not in common:
        common.append(item)
print(common)
```

```python
# 38. Find elements present in the first list but not in the second.
list1 = [10, 20, 30, 40]
list2 = [30, 40, 50, 60]
difference = []
for item in list1:
    if item not in list2:
        difference.append(item)
print(difference)
```

```python
# 39. Rotate a list one position to the right.
items = [1, 2, 3, 4, 5]
rotated = [items[-1]] + items[:-1]
print(rotated)
```

```python
# 40. Rotate a list by a user-given number of positions.
items = [1, 2, 3, 4, 5]
positions = int(input("Enter positions: "))
positions = positions % len(items)
rotated = items[-positions:] + items[:-positions]
print(rotated)
```

```python
# 41. Move all zero values to the end of a list.
items = [0, 5, 0, 3, 8, 0, 2]
non_zero = []
zero_values = []
for item in items:
    if item == 0:
        zero_values.append(item)
    else:
        non_zero.append(item)
result = non_zero + zero_values
print(result)
```

```python
# 42. Divide a list into smaller lists of three elements each.
items = [1, 2, 3, 4, 5, 6, 7, 8]
groups = []
for index in range(0, len(items), 3):
    groups.append(items[index:index + 3])
print(groups)
```

```python
# 43. Check whether a list is a palindrome.
items = [1, 2, 3, 2, 1]
if items == items[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
# 44. Find two numbers whose sum equals a user-given target.
items = [2, 7, 11, 15]
target = int(input("Enter target: "))
found = False
for i in range(len(items)):
    for j in range(i + 1, len(items)):
        if items[i] + items[j] == target:
            print(items[i], items[j])
            found = True
            break
        if found:
            break
if not found:
    print("No pair found")
```
